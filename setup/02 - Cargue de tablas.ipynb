```python
# (df_tejido[cols_expo] > 0).any(axis=1) verifica si hay AL MENOS UN valor > 0 en la fila para esas columnas
df_tejido['HA_EXPORTADO'] = np.where((df_tejido[cols_expo] > 0).any(axis=1), 'Sí', 'No')
```

```python
# Creación de distribución de ingresos

# 1. Cortes dinámicos por deciles
d10 = df_tejido[df_tejido['INGRESOS_OPERACIONALES'] > 0]['INGRESOS_OPERACIONALES'].quantile(0.10)
d20 = df_tejido[df_tejido['INGRESOS_OPERACIONALES'] > 0]['INGRESOS_OPERACIONALES'].quantile(0.20)
d30 = df_tejido[df_tejido['INGRESOS_OPERACIONALES'] > 0]['INGRESOS_OPERACIONALES'].quantile(0.30)
d40 = df_tejido[df_tejido['INGRESOS_OPERACIONALES'] > 0]['INGRESOS_OPERACIONALES'].quantile(0.40)
d50 = df_tejido[df_tejido['INGRESOS_OPERACIONALES'] > 0]['INGRESOS_OPERACIONALES'].quantile(0.50)
d60 = df_tejido[df_tejido['INGRESOS_OPERACIONALES'] > 0]['INGRESOS_OPERACIONALES'].quantile(0.60)
d70 = df_tejido[df_tejido['INGRESOS_OPERACIONALES'] > 0]['INGRESOS_OPERACIONALES'].quantile(0.70)
d80 = df_tejido[df_tejido['INGRESOS_OPERACIONALES'] > 0]['INGRESOS_OPERACIONALES'].quantile(0.80)
d90 = df_tejido[df_tejido['INGRESOS_OPERACIONALES'] > 0]['INGRESOS_OPERACIONALES'].quantile(0.90)

for nombre, valor in [('D10', d10), ('D20', d20), ('D30', d30), ('D40', d40),
                      ('D50', d50), ('D60', d60), ('D70', d70), ('D80', d80), ('D90', d90)]:
    print(f"Corte {nombre}: {valor:,.0f}")
```

```python
# Función auxiliar para que los números grandes se lean mejor en el texto
# Esto convierte 4050000 en "4.1 Millones"
def formato_millones(valor):
    return f"{valor / 1_000_000:,.1f} Millones"
```

```python
# 2. Condiciones lógicas usando los cortes dinámicos por deciles
col = df_tejido['INGRESOS_OPERACIONALES']
condiciones = [
    col == 0,
    (col > 0)   & (col <= d10),
    (col > d10) & (col <= d20),
    (col > d20) & (col <= d30),
    (col > d30) & (col <= d40),
    (col > d40) & (col <= d50),
    (col > d50) & (col <= d60),
    (col > d60) & (col <= d70),
    (col > d70) & (col <= d80),
    (col > d80) & (col <= d90),
    col > d90
]
```

```python
# 3. Textos inyectando las variables calculadas
opciones = [
    'A. COP 0',
    f'B. COP 1 - {formato_millones(d10)}',
    f'C. COP {formato_millones(d10)} - {formato_millones(d20)}',
    f'D. COP {formato_millones(d20)} - {formato_millones(d30)}',
    f'E. COP {formato_millones(d30)} - {formato_millones(d40)}',
    f'F. COP {formato_millones(d40)} - {formato_millones(d50)}',
    f'G. COP {formato_millones(d50)} - {formato_millones(d60)}',
    f'H. COP {formato_millones(d60)} - {formato_millones(d70)}',
    f'I. COP {formato_millones(d70)} - {formato_millones(d80)}',
    f'J. COP {formato_millones(d80)} - {formato_millones(d90)}',
    f'K. Más de COP {formato_millones(d90)}'
]

print(opciones)
```

```python
# 4. Crear columna
df_tejido['RANGO_INGRESOS_TXT'] = np.select(condiciones, opciones, default='L. Información no disponible')
print(df_tejido['RANGO_INGRESOS_TXT'].value_counts())
```

```python
# Lista de pd a subir
bases_de_datos = [
    df_tejido, 
    df_bienes_servicios, 
    df_municipios,
    df_departamentos
]

nombres_tablas = [
    'TEJIDO_EMPRESARIAL_P', 
    'BIENES_Y_SERVICIOS_P', 
    'BASE_MUNICIPIOS_P',
    'CORRELATIVA_DEPARTAMENTOS'
]
```

```python
# Mensaje de inicio de proceso de cargue
print('Iniciando proceso de cargue...')

# pd de verificación
df_resultados_verificacion = pd.DataFrame()
# Subir y verificar bases a Snowflake
for base, tabla in zip(bases_de_datos, nombres_tablas):

    # Cambiar ubicación de la sesión para carga de datos
    snowflake_analitica.update_session_params(sesion_activa, database='APP_SEGMENTACION_EXPORTACIONES', schema='PUBLIC')

    # Obtener números de registros
    obs = len(base)
    
    # Cargar el DataFrame en Snowflake y capturar el mensaje de carga
    mensaje_carga = snowflake_analitica.upload_dataframe_to_snowflake(
            sesion_activa=sesion_activa, 
            df=base, 
            nombre_tabla=tabla, 
            create_table=True, 
            overwrite=True, 
            ram_gb=32
        )
    
    # Verificar y almacenar el resultado en el DataFrame
    resultado = sesion_activa.sql(f"SELECT COUNT(*) FROM {tabla};").collect()
    total_registros = resultado  # Extraer el total de registros

    # Crear un DataFrame temporal para la nueva fila
    nueva_fila = pd.DataFrame({
        'Tabla': [tabla],
        'Total_Registros': [total_registros],
        'Mensaje_Carga': [mensaje_carga]
    })

    # Concatenar la nueva fila al DataFrame de resultados
    df_resultados_verificacion = pd.concat([df_resultados_verificacion, nueva_fila], ignore_index=True)

    # Cambiar ubicación de la sesión para carga de datos a la tabla de auditoria
    snowflake_analitica.update_session_params(sesion_activa,  database='APP_SEGMENTACION_EXPORTACIONES', schema='SEGUIMIENTO')

     # Registrar evento de cargue
    resultado_str = '\n'.join(mensaje_carga)

    snowflake_analitica.registrar_evento_auditoria(sesion_activa=sesion_activa, 
                                                   nombre_esquema_destino='PUBLIC', 
                                                   nombre_tabla=tabla, 
                                                   numero_registros=obs, 
                                                   mensaje=resultado_str)
```

```python
# Query para crear la TABLA que une la base de tejido y municipios
query_tabla_tejido_municipios = """
CREATE OR REPLACE TABLE APP_SEGMENTACION_EXPORTACIONES.SEGMENTACION.TEJIDO_EMPRESARIAL_COMPLETO_BASE_MUNICIPIOS_P AS
SELECT -- 1. BLOQUE DE COLUMNAS ORIGINALES DE TEJIDO_EMPRESARIAL_COMPLETO
        A.NIT,
        A.DIGITO_VERIFICACION,
        A.RAZON_SOCIAL,
        A.TAMANO,
        A.MACRORREGION_EMP,
        A.COD_DEPARTAMENTO_EMP,
        A.DEPARTAMENTO_EMP,
        A.COD_MUNICIPIO_EMP,
        A.MUNICIPIO_EMP,
        A.COD_CIIU_1,
        A.DESCRIPCION_CIIU_1,
        A.CADENA_CIIU_1,
        A.VALOR_AGREGADO_CIIU_1,
        A.CIIU_2,
        A.DESCRIPCION_CIIU_2,
        A.CADENA_CIIU_2,
        A.CIIU_3,
        A.DESCRIPCION_CIIU_3,
        A.CADENA_CIIU_3,
        A.CIIU_4,
        A.DESCRIPCION_CIIU_4,
        A.CADENA_CIIU_4,
        A.ANOS_EMPRESA,
        A.INVERSION_EXTRANJERA,
        A.ACTIVOS,
        A.INGRESOS_OPERACIONALES,
        A.UTILIDAD,
        A.EMPLEADOS,
        A.CANTIDAD_MUJERES_EMPLEADAS,
        A.CANTIDAD_MUJERES_EN_CARGOS_DIRECTIVOS,
        A.CANTIDAD_ESTABLECIMIENTOS,
        A.DIRECCION,
        A.TELEFONO,
        A.EMAIL,
        A.ID_REPRESENTANTE_LEGAL,
        A.REPRESENTANTE_LEGAL,
        A.ORGANIZACION_JURIDICA,
        A.CATEGORIA_MATRICULA,
        A.TIPO,
        A.CADENA,
        A.SECTOR,
        A.SUBSECTOR,
        A.COD_POSICION_ARANCELARIA,
        A.DESC_POSICION_ARANCELARIA,
        A.VALOR_AGREGADO_EXPO,
        A.COD_DEPARTAMENTO_EXPO,
        A.DEPARTAMENTO_EXPO,
        A.PAIS_DESTINO,
        A.HUB,
        A.TRAYECTORIA_EXPORTADORA,
        A.EXPO_2021,
        A.EXPO_2022,
        A.EXPO_2023,
        A.EXPO_2024,
        A.EXPO_2025,
        -- CORRIDO
        A.EXPO_ENE_MAY_2025,
        A.EXPO_ENE_MAY_2026,
        -- CORRIDO
        A.ATENDIDA_PC,
        A.NEGOCIOS,
        A.SERVICIOS,
        A.OPORTUNIDADES,
        A.NUMERO_NEGOCIOS_2023,
        A.NUMERO_NEGOCIOS_2024,
        A.NUMERO_NEGOCIOS_2025,
        -- CORRIDO
        A.NUMERO_NEGOCIOS_ENE_JUN_2026,
        -- CORRIDO
        A.NUMERO_SERVICIOS_2023,
        A.NUMERO_SERVICIOS_2024,
        A.NUMERO_SERVICIOS_2025,
        -- CORRIDO
        A.NUMERO_SERVICIOS_ENE_JUN_2026,
        -- CORRIDO
        A.NUMERO_OPORTUNIDADES_2023,
        A.NUMERO_OPORTUNIDADES_2024,
        A.NUMERO_OPORTUNIDADES_2025,
        -- CORRIDO
        A.NUMERO_OPORTUNIDADES_ENE_JUN_2026,
        -- CORRIDO
        -- COLUMNA EXPORTADORA NME
        A.EXPORTADORA AS EXPORTADORA_NME_ACTIVIDAD,
        -- COLUMNA EXPORTADORA NME
        A.CADENA_SEGMENTACION,
        A.FUENTES,

        -- 2. BLOQUE DE COLUMNAS ORIGINALES DE BASE_MUNICIPIOS_P
        B.MENOR_200K_HABITANTES,
        B.MUNICIPIO AS MUNICIPIO_BASE_MUNICIPIOS,
        B.PDET,
        B.SUBREGION_PDET,
        B._ACT_PRIMARIAS_MUNICIPIO,
        B._ACT_SECUNDARIAS_MUNICIPIO,
        B._ACT_TERCIARIAS_MUNICIPIO,
        B._GRUPOS_ETNICOS_MUNICIPIO,
        B._INFORMALIDAD_MUNICIPIO,
        B._JOVENES_MUNICIPIO,
        B._MUJERES_MUNICIPIO,
        B._POBL_CON_DISCAPACIDAD_MUNICIPIO,
        B._POBL_CON_EDU_TECNICATECNOLOGIA_MUNICIPIO,
        B._POBL_CON_EDUCACION_MEDIA_MUNICIPIO,
        B._POBL_CON_POSGRADO_MUNICIPIO,
        B._POBL_CON_PREGRADO_MUNICIPIO,
        B._POBREZA_MUNICIPIO,
        B.POBLACION_MUNICIPIO,
        B.VALOR_AGREGADO_MUNICIPIO,
        B.ZOMAC,
        B.CODIGO_MUNICIPIO AS CODIGO_MUNICIPIO_BASE_MUNICIPIOS,
        B.CODIGO_DEPARTAMENTO AS CODIGO_DEPARTAMENTO_BASE_MUNICIPIOS,
        B.DEPARTAMENTO AS DEPARTAMENTO_BASE_MUNICIPIOS,
        -- 3. COLUMNAS CALCULADAS

        CASE 
            WHEN A.ANOS_EMPRESA >= 0 AND A.ANOS_EMPRESA <= 5 THEN 'A. 0 - 5 años'
            WHEN A.ANOS_EMPRESA > 5 AND A.ANOS_EMPRESA <= 10 THEN 'B. 6 - 10 años'
            WHEN A.ANOS_EMPRESA > 10 AND A.ANOS_EMPRESA <= 15 THEN 'C. 11 - 15 años'
            WHEN A.ANOS_EMPRESA > 15 AND A.ANOS_EMPRESA <= 20 THEN 'D. 16 - 20 años'
            WHEN A.ANOS_EMPRESA > 20 AND A.ANOS_EMPRESA <= 25 THEN 'E. 21 - 25 años'
            WHEN A.ANOS_EMPRESA > 25 AND A.ANOS_EMPRESA <= 30 THEN 'F. 26 - 30 años'
            WHEN A.ANOS_EMPRESA > 30 AND A.ANOS_EMPRESA <= 35 THEN 'G. 31 - 35 años'
            WHEN A.ANOS_EMPRESA > 35 AND A.ANOS_EMPRESA <= 40 THEN 'H. 36 - 40 años'
            WHEN A.ANOS_EMPRESA > 40 AND A.ANOS_EMPRESA <= 45 THEN 'I. 41 - 45 años'
            WHEN A.ANOS_EMPRESA > 45 AND A.ANOS_EMPRESA <= 50 THEN 'J. 46 - 50 años'
            WHEN A.ANOS_EMPRESA > 50 THEN 'K. Más de 50 años'
            ELSE 'L. Sin clasificar'
        END AS RANGO_ANTIGUEDAD,

        -- COLUMNA DE RANGO DE INGRESOS PREVIAMENTE CALCULADA CON QUANTILES
        A.RANGO_INGRESOS_TXT AS RANGO_INGRESOS,
        -- COLUMNA DE RANGO DE INGRESOS PREVIAMENTE CALCULADA CON QUANTILES

        -- INDICADORA DE EXPORTACIONES MAYORES A CERO EN CUALQUIER AÑO 
        A.HA_EXPORTADO
        -- INDICADORA DE EXPORTACIONES MAYORES A CERO EN CUALQUIER AÑO

FROM APP_SEGMENTACION_EXPORTACIONES.PUBLIC.TEJIDO_EMPRESARIAL_COMPLETO AS A
LEFT JOIN APP_SEGMENTACION_EXPORTACIONES.PUBLIC.BASE_MUNICIPIOS_P AS B ON A.COD_MUNICIPIO_EMP = B.CODIGO_MUNICIPIO;
"""
# Ejecutar
sesion_activa.sql(query_tabla_tejido_municipios).collect()
```

```python
# Verificación de que el left join no aumenta el número de registros en la base de tejido

# Tejido original 
query_count_tejido = "SELECT COUNT(*) FROM APP_SEGMENTACION_EXPORTACIONES.PUBLIC.TEJIDO_EMPRESARIAL_COMPLETO;"

# Ejecutar
count_tejido = sesion_activa.sql(query_count_tejido).collect()

# Tejido con municipios
query_count_tejido_tabla = "SELECT COUNT(*) FROM APP_SEGMENTACION_EXPORTACIONES.SEGMENTACION.TEJIDO_EMPRESARIAL_COMPLETO_BASE_MUNICIPIOS_P;"

# Ejecutar
count_tejido_tabla = sesion_activa.sql(query_count_tejido_tabla).collect()
```

```python
# Verificar cantidad de registros
if count_tejido == count_tejido_tabla:
    # Mensaje de exito
    print("Ambos df tienen el mismo número de filas")
else:
    # Mensaje de error
    print("Error, el número de filas no es igual")
```

```python
# Identificar municipios de tejido que no cruzan
query_municipios_tabla = """
SELECT DISTINCT
    A.COD_MUNICIPIO_EMP,
    A.MUNICIPIO_EMP,
    A.MUNICIPIO_BASE_MUNICIPIOS
FROM APP_SEGMENTACION_EXPORTACIONES.SEGMENTACION.TEJIDO_EMPRESARIAL_COMPLETO_BASE_MUNICIPIOS_P AS A
WHERE A.MUNICIPIO_BASE_MUNICIPIOS IS NULL
ORDER BY A.MUNICIPIO_EMP ASC;
"""

# Ejecutar
df_municipios_no_cruce = pd.DataFrame(sesion_activa.sql(query_municipios_tabla).collect())
df_municipios_no_cruce
```

## 2. Creación de tablas para filtros

```python
# Cambiar ubicación de la sesión para carga de datos
snowflake_analitica.update_session_params(sesion_activa, database='APP_SEGMENTACION_EXPORTACIONES', schema='SEGMENTACION')
```

```python
# Query para filtros generales de para empresas
query_filtros_generales = """
CREATE OR REPLACE TABLE APP_SEGMENTACION_EXPORTACIONES.SEGMENTACION.FILTROS_GENERALES_TEJIDO_EMPRESARIAL_COMPLETO AS
SELECT DISTINCT 
    -- 1. BLOQUE DE COLUMNAS ORIGINALES DE BASE_MUNICIPIOS_P
    COALESCE(A.MENOR_200K_HABITANTES, 'No') AS MENOR_200K_HABITANTES,
    COALESCE(A.PDET, 'No') AS PDET,
    COALESCE(A.DEPARTAMENTO_EMP, 'No determinado') AS DEPARTAMENTO,
    COALESCE(A.MUNICIPIO_EMP, 'No determinado') AS MUNICIPIO,

    -- 2. BLOQUE DE COLUMNAS ORIGINALES DE TEJIDO_EMPRESARIAL_P
    COALESCE(A.TAMANO, 'No determinado') AS TAMANO,
    COALESCE(A.CADENA_SEGMENTACION, 'No determinado') AS CADENA_SEGMENTACION,
    COALESCE(A.TRAYECTORIA_EXPORTADORA, 'No determinado') AS TRAYECTORIA_EXPORTADORA,
    COALESCE(A.COD_CIIU_1, 'No determinado') AS COD_CIIU_1,
    COALESCE(A.DESCRIPCION_CIIU_1, 'No determinado') AS DESCRIPCION_CIIU_1,
    COALESCE(A.VALOR_AGREGADO_CIIU_1, 'No determinado') AS VALOR_AGREGADO_CIIU_1,
    COALESCE(A.CADENA_CIIU_1, 'No determinado') AS CADENA_CIIU_1,
    
    -- 3. COLUMNAS CALCULADAS
    COALESCE(A.RANGO_ANTIGUEDAD, 'No determinado') AS RANGO_ANTIGUEDAD,
    COALESCE(A.RANGO_INGRESOS, 'No determinado') AS RANGO_INGRESOS,
    COALESCE(A.HA_EXPORTADO, 'No determinado') AS HA_EXPORTADO
FROM APP_SEGMENTACION_EXPORTACIONES.SEGMENTACION.TEJIDO_EMPRESARIAL_COMPLETO_BASE_MUNICIPIOS_P AS A;
"""

# Ejecutar
sesion_activa.sql(query_filtros_generales).collect()
```
